```python
# ③ Streamlit をバックグラウンドで起動
import subprocess, threading, time

def _run_streamlit():
    subprocess.run(
        ['streamlit', 'run', '/content/app.py',
         '--server.port', '8501',
         '--server.headless', 'true',
         '--server.enableCORS', 'false',
         '--server.enableXsrfProtection', 'false'],
        capture_output=False
    )

t = threading.Thread(target=_run_streamlit, daemon=True)
t.start()
time.sleep(5)
print('Streamlit 起動完了')
```

```python
# ④ ngrok でトンネルを作成してURLを取得
# ※ ngrok アカウントを持っていない場合は https://ngrok.com で無料登録してトークンを取得
from pyngrok import ngrok

# 初回のみ下記のコメントを外してトークンを設定
# ngrok.set_auth_token("YOUR_NGROK_TOKEN_HERE")

public_url = ngrok.connect(8501)
print(f'\n✅ アプリURL: {public_url}')
print('上記URLをブラウザで開いてください')
```
